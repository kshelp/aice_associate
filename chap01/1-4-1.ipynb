```python
#Json 파일 불러오기
import requests
import pandas as pd

# JSON API 데이터 읽기
url_json = "https://jsonplaceholder.typicode.com/todos"
response = requests.get(url_json)
todos_data = response.json()
df_todos = pd.DataFrame(todos_data)
print(df_todos.head())
```

```python
# SQLite 라이브러리 불러오기
import sqlite3

# SQLite 데이터베이스 연결 및 데이터 읽기
local_file = "../datasets/Chinook_Sqlite.sqlite"  # 불러오는 SQLite 파일 경로
conn = sqlite3.connect(local_file)
query = "SELECT * from Customer"
df_customers = pd.read_sql_query(query, conn)
conn.close()
print(df_customers.head())
```

```python
import pandas as pd
import requests
from io import StringIO

# URL
url_html = "https://en.wikipedia.org/wiki/List_of_countries_by_GDP_(nominal)"

# 헤더 추가
headers = {
    "User-Agent": "Mozilla/5.0"
}

# HTML 가져오기
response = requests.get(url_html, headers=headers)

# HTML 문자열을 StringIO로 감싸기
html_data = StringIO(response.text)

# 테이블 읽기
tables = pd.read_html(html_data)

# GDP 테이블 선택
gdp_table = tables[2]

# 출력
print(gdp_table.head())
```
