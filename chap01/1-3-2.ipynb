```python
import pandas as pd

# Series 생성 
data = [10, 20, 30, 40] 
series = pd.Series(data) 

print("Series:\n", series)
print("첫 번째 값:", series[0])
```

```python
# 딕셔너리를 사용해 DataFrame 생성 
data = { 
    "이름": ["김철수", "이영희", "박민수"], 
    "수학": [85, 90, 78], 
    "영어": [88, 84, 95] 
}

df = pd.DataFrame(data) 
print("DataFrame:\n", df)
```
