#### (1) 데이터 병합

```python
import pandas as pd

# 첫 번째 데이터프레임 생성 
data1 = { 
    '제품ID': [1, 2, 3], 
    '제품명': ['노트북', '모니터', '키보드'] 
} 
df1 = pd.DataFrame(data1)

# 두 번째 데이터프레임 생성 
data2 = { 
    '제품ID': [1, 2, 4], 
    '가격': [1000, 200, 50] 
} 
df2 = pd.DataFrame(data2)

# 데이터프레임 출력 
print("첫 번째 데이터프레임:") 
print(df1) 
print("\n두 번째 데이터프레임:") 
print(df2)
```

```python
# 내부 조인 
inner_join = pd.merge(df1, df2, on='제품ID', how='inner')

# 데이터프레임 출력 
print("첫 번째 데이터프레임:") 
print(df1) 
print("\n두 번째 데이터프레임:") 
print(df2) 
print("\n내부 조인 결과:") 
print(inner_join)
```

#### (2) 데이터 추가(Union)

```python
# 첫 번째 데이터프레임 생성 
data1 = { 
    '이름': ['홍길동', '이순신'], 
    '나이': [30, 40] 
} 
df1 = pd.DataFrame(data1)

# 두 번째 데이터프레임 생성 
data2 = { 
    '이름': ['강감찬', '유관순'], 
    '나이': [50, 20] 
} 
df2 = pd.DataFrame(data2)

# 데이터프레임 출력 
print("첫 번째 데이터프레임:") 
print(df1) 
print("\n두 번째 데이터프레임:") 
print(df2)
```

```python
# 데이터프레임 세로로 결합 
union_df = pd.concat([df1, df2], ignore_index=True)

# 데이터프레임 출력 
print("첫 번째 데이터프레임:") 
print(df1) 
print("\n두 번째 데이터프레임:") 
print(df2) 
print("\n유니온 결과:") 
print(union_df)
```
