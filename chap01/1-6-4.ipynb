#### (1) 시각화 라이브러리

```python
import matplotlib.pyplot as plt 
from matplotlib import rc 
import warnings

# 경고창 제거 
warnings.filterwarnings('ignore')

# 한글 글꼴 설정 
rc('font', family='Malgun Gothic')

# 데이터 준비 
x = [1, 2, 3, 4, 5] 
y = [10, 20, 30, 40, 50]

# 그래프 그리기 
plt.plot(x, y, label="한글 데이터", color='blue', marker='o') 
plt.title("한글과 경고 제거 테스트") # 그래프 제목 
plt.xlabel("X축 이름") # X축 레이블 
plt.ylabel("Y축 이름") # Y축 레이블 
plt.legend() # 범례 표시 
plt.grid(True) # 격자 표시 
plt.show()
```

```python
import seaborn as sns 
import pandas as pd

# 샘플 데이터프레임 생성 
data = pd.DataFrame({ 
    'Category': ['A', 'B', 'C', 'D'], 
    'Values': [5, 7, 3, 8] 
})

# 막대 그래프 
sns.barplot(x='Category', y='Values', data=data) 
plt.title("Bar Plot Example") 
plt.show()
```

#### (2) 그래프 그리기 기본 코드

##### matplotlib.pyplot: 가장 기본적인 시각화 도구

```python
# 시각화 라이브러리 임포트 
# 시각화 출력시 경고메시지 안나오는 코드
import matplotlib.pyplot as plt 
import matplotlib.font_manager as fm
import seaborn as sns 
import warnings 

warnings.filterwarnings('ignore')

# 한글 폰트 설정 (Windows 기준: 'Malgun Gothic') 
plt.rcParams['font.family'] = 'Malgun Gothic'

# 마이너스 부호 깨짐 방지 
plt.rcParams['axes.unicode_minus'] = False
```

```python
# 데이터셋 임포트 
df = pd.read_csv("../datasets/heart.csv")

# 1. 나이별 평균 콜레스테롤 계산 
age_chol = df.groupby('age')['chol'].mean().reset_index()

# 2. 선 그래프 그리기 
plt.plot(age_chol['age'], age_chol['chol']) 
plt.xlabel('나이') 
plt.ylabel('평균 콜레스테롤 수치') 
plt.title('나이별 평균 콜레스테롤 변화') 
plt.grid(True) 
plt.show() 
```

```python
plt.bar(df['sex'], df['chol']) # x축: 성별(0,1), y축: 평균 콜레스테롤 
plt.xlabel('성별') 
plt.ylabel('콜레스테롤 수치') 
plt.title('성별에 따른 콜레스테롤 비교') 
plt.show()
```

##### seaborn: 더 예쁜 그래프, 통계 기반 시각화

```python
# 시각화 라이브러리 임포트 
# 시각화 출력시 경고메시지 안나오는 코드 
import matplotlib.pyplot as plt 
import matplotlib.font_manager as fm
import seaborn as sns
import warnings 

warnings.filterwarnings('ignore')

# 한글 폰트 설정 (Windows 기준: 'Malgun Gothic') 
plt.rcParams['font.family'] = 'Malgun Gothic'

# 마이너스 부호 깨짐 방지 
# plt.rcParams['axes.unicode_minus'] = False
```

```python
import seaborn as sns

sns.lineplot(data=df, x='age', y='chol') 
plt.title('나이에 따른 콜레스테롤 변화') 
plt.show()
```

```python
sns.barplot(data=df, x='sex', y='chol') # 평균 자동 계산 
plt.title('성별에 따른 평균 콜레스테롤') 
plt.show()
```

```python
sns.boxplot(data=df, x='sex', y='thalach')
plt.title('성별에 따른 심박수 분포 (이상치 포함)') 
plt.show()
```

#### (4) 박스 그래프(box plot)

```python
import pandas as pd 
import matplotlib.pyplot as plt

# CSV 데이터 읽기 
data = pd.read_csv("../datasets/heart.csv")

# 박스 그래프 생성 
plt.boxplot( 
    data['age'], # 나이 데이터를 시각화 
    vert=True, # 그래프를 세로로 표시 
    patch_artist=True, # 박스 내부 색상 채우기 
    boxprops=dict(facecolor='skyblue', color='black'), # 박스 스타일 
    medianprops=dict(color='red', linewidth=2), # 중앙값 스타일 
    whiskerprops=dict(color='black', linewidth=1.5), # 수염 스타일 
    capprops=dict(color='black', linewidth=1.5) # 상한/하한선 스타일 
) 
plt.title("Age Distribution in Heart Dataset") # 그래프 제목 설정 
plt.ylabel("Age") # Y축 이름 설정 
plt.grid(True) # 격자 표시 
plt.show()
```
