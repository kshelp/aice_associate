```python
# 필요한 라이브러리 불러오기
import numpy as np
from tensorflow.keras.datasets import mnist # Keras에서 MNIST 데이터셋 로드
from sklearn.model_selection import train_test_split # 데이터 분리
from sklearn.ensemble import GradientBoostingClassifier # 그래디언트 부스팅 모델
from sklearn.metrics import accuracy_score, classification_report # 평가 지표
import matplotlib.pyplot as plt # 시각화
from tqdm import tqdm
import joblib
```

```python
# 1. 데이터 로드 및 전처리
(X_train_full, y_train_full), (X_test, y_test) = mnist.load_data()
X_train_full = X_train_full / 255.0
X_test = X_test / 255.0
X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full, test_size=0.2, random_state=42)
X_train = X_train.reshape(X_train.shape[0], -1)
X_val = X_val.reshape(X_val.shape[0], -1)
X_test = X_test.reshape(X_test.shape[0], -1)
```

```python
# 2. 그래디언트 부스팅 모델 생성 및 학습
gb_model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42)

# 학습 진행 상황 표시
for i in tqdm(range(1, gb_model.n_estimators + 1), desc="Training Progress"):
    gb_model.n_estimators = i
    gb_model.fit(X_train, y_train)

# 학습된 모델 저장
loaded_model = joblib.load("gradient_boosting_model.pkl")
```

```python
# 3. 테스트 데이터로 예측
# 저장된 모델 불러오기
loaded_model = joblib.load("gradient_boosting_model.pkl")
y_pred = loaded_model.predict(X_test)
```

```python
# 4. 모델 평가
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)
print("Gradient Boosting Accuracy:", accuracy)
print("\nClassification Report:\n", report)
```
